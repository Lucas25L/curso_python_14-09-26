## Buscar datos del precio de las acciones

```python
%pip install yfinance
%pip install matplotlib
```

```python
import yfinance as yf
```

```python
ticker = input("Ingrese el ticker de la acción: ")

data = yf.download([ticker], period="6mo")

cierre = data.Close[ticker]
```

```python
cierre.plot()
```

## Hacer los analisis solicitados por el gestor
* Cotización máxima
* Cotización mínima
* Valor medio

```python
maxima = round(cierre.max(), 2)
minima = round(cierre.min(), 2)
promedio = round(cierre.mean(), 2)
```

## Enviar mail

```python
%pip install pyautogui
%pip install pyperclip
```

```python
import pyautogui
import pyperclip
import webbrowser
import time
```

```python
webbrowser.open("https://mail.google.com/mail/u/0/#inbox")

destinatario = "ejemplo@gmail.com"
asunto = "Análisis de la acción " + ticker + "en los últimos 6 meses"
cuerpo = f"""
Buenos días estimado, le informo que el análisis de la acción {ticker} en los últimos 6 meses es el siguiente:
- Precio máximo: USD {maxima}
- Precio mínimo: USD {minima}
- Precio promedio: USD {promedio}
Estoy a su disposición para cualquier consulta adicional.
Saludos cordiales,
Lucas
"""

# Esperar 3 segundos para que se abra el navegador
time.sleep(7)

# Configurar una pausa de 3 segundos entre cada acción de PyAutoGUI
pyautogui.PAUSE = 7

# Hacer clic en el botón "Redactar"
pyautogui.click(92, 242)

pyperclip.copy(destinatario)
pyautogui.hotkey("ctrl", "v")
pyautogui.press("tab")

pyperclip.copy(asunto)
pyautogui.hotkey("ctrl", "v")
pyautogui.press("tab")

pyperclip.copy(cuerpo)
pyautogui.hotkey("ctrl", "v")

pyautogui.click(846, 705)  # Hacer clic en el botón "Enviar"

# Cerrar la pestaña del navegador una vez finalizado el envío del correo
pyautogui.hotkey("ctrl", "w")
```
